Step 5 : use a function that uses random mod to randomly decide which player goes first using randint

```python
import random
def choose_first():
  flip = random.randint(0,1)
  if flip == 0:
    return 'player1'
  else:
    return 'player2'
```

Step 6 : write a function that returns a boolean indicating whether a space on the board is freely available

```python
def space_check(board, position):
  return board[position]==" "
```

Step 7: write a fn that checks if the board is full or not if full returns true else returns false

```python
def full_board_check(board):
  for i in range(1,10):
     if space_check(board,i):
      return False
  #board is full, we return true
  return True
```

Step 8 : write a function that asks for players next position and then uses funtion from step 6 to check if its a free psotion or not, if it is then return the position for later use

```python
def player_choice(board):
  position = 0

  while position not in [1,2,3,4,5,6,7,8,9] or not space_check(board,position):
    position = int(input("Choose a position:(1-9) "))
  return position
```
