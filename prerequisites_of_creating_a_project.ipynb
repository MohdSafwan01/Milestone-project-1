lets fix these issues

for fixiing the fist issue we will use isdigit() method

```python
def user_choice():
  choice = 'WRONG'

  while choice.isdigit() == False:

    choice = input('Please enter a number(0-10):')

    if choice.isdigit()==False:
      print('please enter a digit')

  return int(choice)
```

Now lets fix the range issue

```python
def user_choice():
#initials
 choice = 'WRONG'
 acceptable_range = range(0,10)
 within_range=False

 #Two conditions need to be proved
 #choice.isdigit() and within range == False

 while choice.isdigit() == False or within_range == False:

  choice = input('please enter a value(0-10): ')

  #check for digit:
  if choice.isdigit() == False:
    print("Please enter a digit")

  #check for range:
  if choice.isdigit()==True:
    if int(choice) in acceptable_range:
      within_range = True
    else:
      within_range = False
```
