Here we are creating a simle user interaction game

by using all the previous methons and tricks used

```python
game_list=[0,1,2]
```

displaying the gamelist

```python
def display_game(game_list):
  print('here is the current list: ')
  print(game_list)
```

```python
display_game(game_list)
```

choosing up the position

```python
def position_choice():
   choice = 'WRONG'
   while choice not in ['0','1','2']:
     choice = input('choose a position (0,1,2) :')
     if choice not in ['0','1','2']:
      print("Sorry, invalid choice")
   return int(choice)
```

```python
position_choice()
```

Replacing position with string

```python
def replacement_choice(game_list,position):
  user_placement = input("Type a string to replace at position: ")

  game_list[position] = user_placement
  return game_list
```

```python
replacement_choice(game_list,2)
```

deciding wheter the game is on or off

```python
def gameon_choice():
  choice = 'Wrong'

  while choice not in ['Y','N']:
    choice = input("Keep playing (Y or N))")

    if choice not in ['Y','N']:
      print("Sorry, i dont understand,please choose Y or N")

  if choice =='Y':
      return True
  else:
      return False
```

```python
gameon_choice()
```

```python
game_on = True
game_list = [0,1,2]

while game_on:
  display_game(game_list)

  position = position_choice()

  game_list=replacement_choice(game_list,position)

  display_game(game_list)

  game_on = gameon_choice()
```
